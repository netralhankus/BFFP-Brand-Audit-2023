# Menjadikan baris bertama sebagai nama kolom

```python
df.columns = df.iloc[0]
df = df[1:]
df = df.reset_index(drop=True)
df
```

```python
df.columns
```

# Menghapus kolom yang tidak diperlukan

```python
df = df.drop(columns=['Item Description', 'Type of Audit'])
display(df.head())
```

#Menggabungkan data parent company dari dua kolom

```python
df.columns
```

```python
df['Parent Company'] = df['Parent Company'].fillna(df['If Parent Company does not appear automatically or needs correction, please add here.'])
df['Parent Company'] = df['Parent Company'].fillna('Unknown')
df = df.drop(columns=['If Parent Company does not appear automatically or needs correction, please add here.'])

current_columns = df.columns.tolist()
if 'Parent Company' in current_columns:
    current_columns.remove('Parent Company')

brand_name_index = current_columns.index('Brand Name')
new_column_order = current_columns[:brand_name_index + 1] + ['Parent Company'] + current_columns[brand_name_index + 1:]
df = df[new_column_order]
display(df.head())
```

```python
df['Parent Company']
```

```python
df['Parent Company'].unique()
```

#Cleaning kolom layers

```python
df['Layers'].unique()
```

```python
filtered_df = df[df['Layers'] == 'SL ']
print(f"\nJumlah 'Layers' dengan 'SL ': {len(filtered_df)}")
display(filtered_df.head())
```

```python
df['Layers'] = df['Layers'].replace(['SL', 'SL '], 'Single Layer')
df['Layers'] = df['Layers'].replace('ML', 'Multi Layer')

print("Unique values di 'Layers':", df['Layers'].unique())
display(df.head())
```

#Cleaning kolom type of material

```python
df['Type of Material'].unique()
```

```python
nan_materials = df[df['Type of Material'].isna()]
print(f"Jumlah data NaN di 'Type of Material': {len(nan_materials)}")
display(nan_materials)
```

```python
df['Type of Material'] = df['Type of Material'].fillna('Others')
df['Type of Material'] = df['Type of Material'].replace('O', 'Others')

print("Unique values di 'Type of Material':", df['Type of Material'].unique())
display(df.head())
```

#Cleaning kolom type of product

```python
df['Type of Product'].unique()
```

```python
df['Type of Product'] = df['Type of Product'].replace(['FP','FP '], 'Food Packaging')
df['Type of Product'] = df['Type of Product'].replace('SM', 'Smoking Material')
df['Type of Product'] = df['Type of Product'].replace('PC', 'Personal Care')
df['Type of Product'] = df['Type of Product'].replace('HP', 'Household Product')
df['Type of Product'] = df['Type of Product'].replace('O', 'Others')
df['Type of Product'] = df['Type of Product'].replace('FG', 'Fishing Gear')
df['Type of Product'] = df['Type of Product'].replace('PM', 'Packaging Material')

print("Unique values di 'Type of Product':", df['Type of Product'].unique())
display(df.head())
```

#Membuat semua Uppercase

```python
# Iterate through all columns in the DataFrame
for col in df.columns:
    # Check if the column's dtype is 'object' (which typically means strings or mixed types)
    if df[col].dtype == 'object':
        # Convert column to string type and then apply .str.upper() for robustness
        # This handles cases where 'object' dtype might contain non-string elements
        df[col] = df[col].astype(str).str.upper()

display(df.head())
```

```python
df.info()
```

```python
import numpy as np
df['Total Count'] = df['Total Count'].astype(str).str.extract('(\d+)').astype(float)
df['Total Count'] = df['Total Count'].astype('Int64')
df['Time of Audit'] = pd.to_datetime(df['Time of Audit']).dt.date

for col in df.columns:
    if col not in ['Total Count', 'Time of Audit']:
        df[col] = df[col].astype(str)

df.info()
display(df.head())
```
